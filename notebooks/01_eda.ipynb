```python
# BLOCK 1 — Import your tools
import pandas as pd           # pd = short nickname. Handles tables of data
import matplotlib.pyplot as plt  # plt = for drawing charts
import seaborn as sns         # sns = prettier charts built on matplotlib
```

```python
# BLOCK 2 — Load the dataset
df = pd.read_csv('../data/raw/zomato.csv',
                 encoding='latin-1')  # encoding fixes special characters
# df = "dataframe" = your table of data in memory
```

```python
# BLOCK 3 — First look at the data
print(df.shape)        # How many rows and columns? e.g. (51717, 17)
print(df.columns)      # What are all the column names?
df.head()              # Show the first 5 rows as a table
```

```python
# BLOCK 4 — Check for problems
print(df.isnull().sum())  # Count missing values per column
df.info()              # Data types of each column (text vs number)
```

```python
# BLOCK 5 — Basic stats
df.describe()          # Min, max, average for number columns
```

```python
# BLOCK 6 — Top 10 restaurant types
top_types = df['rest_type'].value_counts().head(10)
print(top_types)       # Which restaurant type appears most?
```

```python
# BLOCK 7 — Simple bar chart
plt.figure(figsize=(10, 5))          # Set chart size (width, height in inches)
top_types.plot(kind='bar', color='coral')  # Draw the bars
plt.title('Top 10 Restaurant Types in Bangalore')  # Chart title
plt.xlabel('Restaurant Type')        # X axis label
plt.ylabel('Count')                  # Y axis label
plt.xticks(rotation=45)              # Tilt x labels so they fit
plt.tight_layout()                   # Fix spacing so nothing is cut off
plt.savefig('../visuals/charts/top_rest_types.png')  # Save it!
plt.show()    
```
